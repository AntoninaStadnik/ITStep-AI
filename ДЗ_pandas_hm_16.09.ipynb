# Завдання 6
Перевірте чи правильно розраховано загальну вартість покупки. Виведіть ті рядки де є помилка. Виправте дані в таблиці

```python
mask_error = df['sale_amt'] != df['units'] * df['unit_price']

df[mask_error]
```

```python
df.loc[mask_error, 'sale_amt'] = df.loc[mask_error, 'units'] * df.loc[mask_error, 'unit_price']
```
